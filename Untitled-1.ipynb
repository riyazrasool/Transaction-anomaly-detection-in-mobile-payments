```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv("final_4M_upi_fraud.csv")

print("Shape:", df.shape)
df.head()
```

```python
# Normalize drivers
v = df["s_txn_count_10min"] / (df["s_txn_count_10min"].max() + 1)
r = df["r_unique_senders_24h"] / (df["r_unique_senders_24h"].max() + 1)
a = np.log1p(df["amount"]) / np.log1p(df["amount"].max() + 1)
g = df["geo_distance_km"] / (df["geo_distance_km"].max() + 1)
n = df["vpn_proxy_flag"]

# Fraud intent signal (90%)
signal = (
    0.45 * v +
    0.45 * r +
    0.2 * a +
    0.1 * g +
    0.1* n
)

# Light noise (10%)
noise = np.random.normal(0, 0.03, size=len(df))

fraud_score = signal + noise

threshold = np.quantile(fraud_score, 0.8)
df["is_fraud"] = (fraud_score > threshold).astype(int)
```

```python
df["is_fraud"].mean()
```

```python
X = df.drop("is_fraud", axis=1)
y = df["is_fraud"]
```

```python
skewed = [
    "amount",
    "s_avg_amount_5",
    "s_max_amount_5",
    "s_amount_std_5",
    "r_amount_sum_24h",
    "geo_distance_km"
]

for col in skewed:
    X[col] = np.log1p(X[col])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scale_cols = [
    "amount","s_avg_amount_5","s_max_amount_5","s_amount_std_5",
    "s_time_gap_avg","r_amount_sum_24h","geo_distance_km",
    "s_txn_count_10min","s_txn_count_1h","r_txn_count_1h",
    "r_txn_count_24h","r_unique_senders_24h"
]

X[scale_cols] = scaler.fit_transform(X[scale_cols])
```

```python
split = int(0.7 * len(X))
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]
```

```python
import lightgbm as lgb
from sklearn.metrics import average_precision_score

train_data = lgb.Dataset(X_train, label=y_train)
valid_data = lgb.Dataset(X_test, label=y_test)

params = {
    "objective": "binary",
    "metric": "binary_logloss",
    "boosting_type": "gbdt",
    "learning_rate": 0.05,
    "num_leaves": 64,
    "max_depth": -1,
    "min_data_in_leaf": 200,
    "feature_fraction": 0.8,
    "bagging_fraction": 0.8,
    "bagging_freq": 5,
    "scale_pos_weight": (len(y_train)-sum(y_train)) / sum(y_train)
}

model = lgb.train(
    params,
    train_data,
    num_boost_round=500,
    valid_sets=[valid_data],
    valid_names=['validation'],
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(50)]
)
```

```python
probs = model.predict(X_test)
preds = (probs > 0.98).astype(int)
pr_auc = average_precision_score(y_test, probs)

print("PR-AUC:", pr_auc)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, preds))
```

```python
from scipy.stats import ks_2samp

ks = ks_2samp(probs[y_test==1], probs[y_test==0]).statistic
print("KS Statistic:", ks)
```

```python
blocked = preds.sum()
caught = ((preds==1)&(y_test==1)).sum()


print("Blocked txns:", blocked)
print("Fraud caught:", caught)
```

```python
import joblib

joblib.dump(model, "fraud_engine.pkl")
joblib.dump(scaler, "scaler.pkl")
```

```python
from sklearn.metrics import confusion_matrix, roc_curve, auc, precision_recall_curve, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (16, 12)

fig = plt.figure(figsize=(16, 12))

# 1. Confusion Matrix Heatmap
ax1 = plt.subplot(2, 3, 1)
cm = confusion_matrix(y_test, preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax1, cbar=True,
            xticklabels=['Legitimate', 'Fraud'], yticklabels=['Legitimate', 'Fraud'])
ax1.set_title('Confusion Matrix', fontsize=14, fontweight='bold')
ax1.set_ylabel('Actual')
ax1.set_xlabel('Predicted')

# 2. Blocked vs Caught Transactions
ax2 = plt.subplot(2, 3, 2)
categories = ['Total Blocked', 'Fraud Caught', 'False Positives']
values = [blocked, caught, blocked - caught]
colors = ['#FF6B6B', '#4ECDC4', '#FFE66D']
bars = ax2.bar(categories, values, color=colors, edgecolor='black', linewidth=1.5)
ax2.set_title('Transaction Blocking Analysis', fontsize=14, fontweight='bold')
ax2.set_ylabel('Count')
for bar, val in zip(bars, values):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(val)}', ha='center', va='bottom', fontweight='bold')

# 3. ROC Curve
ax3 = plt.subplot(2, 3, 3)
fpr, tpr, _ = roc_curve(y_test, probs)
roc_auc = auc(fpr, tpr)
ax3.plot(fpr, tpr, color='#FF6B6B', lw=2.5, label=f'ROC (AUC = {roc_auc:.3f})')
ax3.plot([0, 1], [0, 1], color='gray', lw=2, linestyle='--', label='Random Classifier')
ax3.set_xlim([0.0, 1.0])
ax3.set_ylim([0.0, 1.05])
ax3.set_xlabel('False Positive Rate')
ax3.set_ylabel('True Positive Rate')
ax3.set_title('ROC Curve', fontsize=14, fontweight='bold')
ax3.legend(loc="lower right")
ax3.grid(True, alpha=0.3)

# 4. Precision-Recall Curve
ax4 = plt.subplot(2, 3, 4)
precision, recall, _ = precision_recall_curve(y_test, probs)
ax4.plot(recall, precision, color='#4ECDC4', lw=2.5, label=f'PR-AUC = {pr_auc:.3f}')
ax4.fill_between(recall, precision, alpha=0.2, color='#4ECDC4')
ax4.set_xlabel('Recall')
ax4.set_ylabel('Precision')
ax4.set_title('Precision-Recall Curve', fontsize=14, fontweight='bold')
ax4.legend(loc="best")
ax4.grid(True, alpha=0.3)

# 5. Prediction Score Distribution
ax5 = plt.subplot(2, 3, 5)
ax5.hist(probs[y_test==0], bins=50, alpha=0.6, label='Legitimate', color='#4ECDC4', edgecolor='black')
ax5.hist(probs[y_test==1], bins=50, alpha=0.6, label='Fraud', color='#FF6B6B', edgecolor='black')
ax5.axvline(0.5, color='black', linestyle='--', linewidth=2, label='Decision Threshold')
ax5.set_xlabel('Prediction Score')
ax5.set_ylabel('Frequency')
ax5.set_title('Score Distribution by Class', fontsize=14, fontweight='bold')
ax5.legend()
ax5.grid(True, alpha=0.3, axis='y')

# 6. Performance Metrics Summary
ax6 = plt.subplot(2, 3, 6)
ax6.axis('off')

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

tn, fp, fn, tp = cm.ravel()
accuracy = accuracy_score(y_test, preds)
precision = precision_score(y_test, preds)
recall = recall_score(y_test, preds)
f1 = f1_score(y_test, preds)
specificity = tn / (tn + fp)

metrics_text = f"""
PERFORMANCE METRICS

Accuracy:     {accuracy:.4f}
Precision:    {precision:.4f}
Recall:       {recall:.4f}
Specificity:  {specificity:.4f}
F1-Score:     {f1:.4f}
PR-AUC:       {pr_auc:.4f}
ROC-AUC:      {roc_auc:.4f}
KS Statistic: {ks:.4f}

CONFUSION MATRIX
TP: {tp}  |  FP: {fp}
FN: {fn}  |  TN: {tn}

BLOCKING STATS
Total Blocked: {blocked}
Fraud Caught:  {caught}
False Alarms:  {fp}
"""

ax6.text(0.1, 0.5, metrics_text, fontsize=11, verticalalignment='center',
         fontfamily='monospace', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))

plt.tight_layout()
plt.show()

print("✓ Comprehensive fraud detection analysis completed")
```
